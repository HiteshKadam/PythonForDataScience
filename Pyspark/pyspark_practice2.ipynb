#### Pyspark Handling Missing Values
- Dropping Columns
- Dropping Rows
- Various Parameter In Dropping functionalities
- Handling Missing values by Mean, MEdian And Mode

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Learn Spark").getOrCreate()
```

```python
df_spark = spark.read.csv("test2.csv", header=True, inferSchema=True)
```

```python
df_spark.show()
```

```python
df_spark.show()
```

```python
df_spark.na.drop().show()
```

```python
df_spark.na.drop(how="any",thresh=2).show()
```

```python
df_spark.na.drop(how="any",subset=["Age"]).show()
```

```python
df_spark.na.fill("Missing Value").show()
```

```python
from pyspark.ml.feature import Imputer

imputer = Imputer(
    inputCols=['age', 'Experience', 'Salary'], 
    outputCols=["{}_imputed".format(c) for c in ['age', 'Experience', 'Salary']]
    ).setStrategy("mean")
```
