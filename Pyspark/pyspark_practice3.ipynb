```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder.appName('dataframe').getOrCreate()
```

```python
df_spark = spark.read.csv('test1.csv', inferSchema=True, header=True)
df_spark.show()
```

#### Filter O[perations]

```python
df_spark.filter("Salary<=20000").show()
```

```python
df_spark.filter("Salary<=20000").select(['Name', 'Salary']).show()
```

```python
df_spark.filter((df_spark['Salary'] <=20000) & (df_spark['Age'] >= 25)).select(['Name', 'Salary']).show()
```
