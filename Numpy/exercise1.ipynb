##### arange(start,stop,step)  You know the step size
##### linspace(start,stop,step) You know how many values you need
##### logspace(start,stop,step) You need logarithmic spacing (10^0 10^1 10^2)

```python
np.arange(5,11)
```

```python
np.arange(0, 30, 5)
```

```python
np.linspace(0, 100, 21)
```

```python
np.logspace(0, 5,5)
```
