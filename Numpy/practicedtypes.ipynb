# Initialize Different types of arrays

```python
import numpy as np
```

```python
#full zeros
np.zeros((2,3))
```

```python
np.ones((4,2,2), dtype='int32')
```

```python
a = np.full((2,2),99)
print(a)
```

```python
# any other number full_like shape of a
np.full_like(a,4)
```

```python
#random numbers default decimals
np.random.rand(4,2)
```

```python
#random integer from 0-7
np.random.randint(0,8,size=(3,3))
```

```python
#identity matrix
np.identity(5)
```

```python
arr = np.array([[1,2,3]])
ri = np.repeat(arr,3,axis=0)
print(ri)
```

```python
output = np.ones((5,5))
print(output)

zeros = np.zeros((3,3))
zeros[1,1] = 9
print(zeros)

output[1:4,1:4] = zeros
print(output)
```

#### Be careful with copying arrays

```python
a = np.array([1,2,3])
b = a.copy()
b[0] = 100

print(a)
print(b)
```

# Linear Algebra

```python
a = np.full((2,3),1)
print(a)

b = np.full((3,2),2)
print(b)
```

```python
np.matmul(a,b)
```

# Reorganizing array

```python
before = np.array([[1,2,3],[4,5,6]])
after = before.reshape((3,2))
print(before)
print(after)
```

```python
# vertical stack
v1 = np.array([1,2,3,4])
v2 = np.array([5,6,7,8])
np.vstack([v1,v2])
```

```python
#horizontal stack
h1 = np.ones((2,4))
h2 = np.zeros((2,2))
np.hstack((h1,h2))
```
