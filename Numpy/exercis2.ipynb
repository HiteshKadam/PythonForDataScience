```python
arr = np.arange(1,21)
print(arr[arr > 10])
```

```python
print(arr[arr % 2 == 0])
```

```python
arr = np.array([8,3,9,1,6])
print(np.sort(arr)[::-1])
print(np.sort(arr))
```

```python
print(np.sort(arr)[-3:])
```

```python
print(np.argsort(arr))
```

```python
cities = np.array([
    "Delhi",
    "Mumbai",
    "Pune",
    "Bengaluru"
])

population = np.array([
    20,
    12,
    7,
    13
])

cities[population.argsort()]
```

```python
scores = np.array([75,92,81,68,99,88])
print(scores[scores.argsort()[-3:]])
```
