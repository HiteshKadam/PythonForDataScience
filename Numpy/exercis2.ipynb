```python
students = np.array([
    "Alice",
    "Bob",
    "Charlie",
    "David",
    "Emma"
])

marks = np.array([
    92,
    74,
    61,
    45,
    88
])


grades = np.select(conditions, choices, default="D")
labels = np.where(marks > 50,"Pass","Fail")
bin_labels = np.where(labels == "Pass", 1, 0)

np.array([students,marks,grades,labels,bin_labels]).T
```
